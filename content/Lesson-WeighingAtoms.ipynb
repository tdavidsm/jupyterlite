# Pretending to Weigh Atoms One At a Time

So, what would happen if we pretended to weigh atoms one at a time? Well, we'd start with a pile of atoms that we didn't know the weight of, let's say gallium atoms because it is going to help us see the interesting things for both atoms and estimating. Hit the play button on the code below to start running the simulation. Weigh at least 10 atoms to be sure of your results.

```python
'''Code was originally generated by AI, and it didn't work (at all
at first and then just not as requested/intended). I have vigorously
massaged it into shape so that it now does what I want!'''

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import IPython
from ipywidgets import Button, Output, VBox, HBox, Layout
from IPython.display import display

# Define isotopic masses and natural abundances for Gallium
gallium_isotopes = {
    'Ga-69': {'mass': 68.925573, 'abundance': 0.60108},
    'Ga-71': {'mass': 70.924705, 'abundance': 0.39892}
}

def weigh_atom():
  """Simulates weighing a single gallium atom.

  Randomly selects an isotope based on natural abundance and returns its mass.

  Returns:
      float: The mass of the selected gallium isotope.
  """
  isotopes = list(gallium_isotopes.keys())
  abundances = [isotope_data['abundance'] for isotope_data in gallium_isotopes.values()]
  chosen_isotope = np.random.choice(isotopes, p=abundances)
  return gallium_isotopes[chosen_isotope]['mass']

# 1. Create an empty pandas DataFrame
df_weighings = pd.DataFrame(columns=['Trial','Element','Mass'])

output_area = Output()
output_area2 = Output()

def update_simulation():
  """Simulates weighing one atom, updates the DataFrame, and refreshes the plot."""
  global df_weighings, output_area2 # Declare variables as global to modify them

  # 2. Call the weigh_atom() function
  atom_mass = weigh_atom()

  # 3. Determine the next trial number
  next_trial = len(df_weighings) + 1

  # 4. Create a new row and store it. (I had to fix AI's code for 4&5)
  df_weighings.loc[len(df_weighings)] = (next_trial, 'Ga', atom_mass)

  plt.scatter(df_weighings['Trial'], df_weighings['Mass'])
  plt.xlabel('Trial Number')
  plt.ylabel('Mass (amu)')
  plt.title('Simulated Weighing of Individual Gallium Atoms')
  plt.ylim(68,72)
  plt.xlim(0,len(df_weighings)+1)


  # 10. Display the updated df_weighings DataFrame
  plt.show()
  with output_area2:
    output_area2.clear_output(wait=True) # Clear previous output before updating
    display(df_weighings)




# 2. Create a Button instance
weigh_button = Button(description='Weigh One Atom with Tweezers', layout=Layout(width='Auto'))

# 4. Define a function to be executed when the button is clicked
def on_button_clicked(b):
    with output_area:
        output_area.clear_output(wait=True) # Clear previous output before updating
        update_simulation() # Call the function to update simulation

# 5. Link the button's on_click event
weigh_button.on_click(on_button_clicked)

# 6. Display the button and the output widget
display(HBox([weigh_button, output_area2, output_area]))
```

**What Happened?**
That was weird and annoying. The atoms weighed different amounts. They were all gallium atoms, but something was different about them. They seemed to be two flavors of the same element. This is because atoms of gallium do come in two different "isotopes." That means there are two different arrangements of protons and neutrons that make stable gallium atoms. One of those arrangements weighs in at 68.925573, and the other one weighs in at 70.924705. So what are we supposed to do when we want to weigh out some gallium to use in microcips or something? How will we know what kind we're getting? Do we have to look at ever single atom? There's no way that's how we do things in the real world.

# Pretending to Weigh Scoops of Atoms

**If It's Good Enough for Ice Cream...**

So, to get a handle on this situation, we're going to do the same sort of simulation, but we're going to use some small scoops of a few atoms at a time and see what kind of results we get. Press play on the code below and gather up some data about delicious scoops of ice cream, I mean atoms.

```python
def weigh_scoop_of_atoms():
  """Simulates weighing a scoop of gallium atoms.
  Randomly selects how many atoms to weigh.
  Randomly selects an isotope for each atom based on natural abundance.
  Sums the total mass of all scooped atoms.

  Returns:
      [int, float]: [The number of atoms, Their total mass all together]
  """
  num_atoms = np.random.randint(1, 6)
  isotopes = list(gallium_isotopes.keys())
  abundances = [isotope_data['abundance'] for isotope_data in gallium_isotopes.values()]
  total_mass = 0
  for _ in range(num_atoms):
    chosen_isotope = np.random.choice(isotopes, p=abundances)
    total_mass += gallium_isotopes[chosen_isotope]['mass']
  return [num_atoms,total_mass]

# 1. Create an empty pandas DataFrame
df_weighings = pd.DataFrame(columns=['Trial','Element','Atoms','Mass'])

output_area = Output()
output_area2 = Output()

def update_simulation():
  """Simulates weighing one atom, updates the DataFrame, and refreshes the plot."""
  global df_weighings, output_area2 # Declare variables as global to modify them

  # 2. Call the weigh_atom() function
  trial_results = weigh_scoop_of_atoms()

  # 3. Determine the next trial number
  next_trial = len(df_weighings) + 1

  # 4. Create a new row and store it. (I had to fix AI's code for 4&5)
  df_weighings.loc[len(df_weighings)] = (next_trial, 'Ga', trial_results[0], trial_results[1])

  plt.scatter(df_weighings['Atoms'], df_weighings['Mass'], marker='+')
  plt.xlabel('Number of Atoms in Scoop')
  plt.ylabel('Mass (amu)')
  plt.title('Simulated Weighing of Scoops of Gallium Atoms')
  plt.ylim(0,df_weighings['Mass'].max()+0.1*df_weighings['Mass'].max())
  plt.xlim(0,df_weighings['Atoms'].max()+.1*df_weighings['Atoms'].max())


  # 10. Display the updated df_weighings DataFrame
  plt.show()
  with output_area2:
    output_area2.clear_output(wait=True) # Clear previous output before updating
    display(df_weighings)




# 2. Create a Button instance
weigh_button = Button(description='Weigh a Small Scoop of Atoms', layout=Layout(width='Auto'))

# 4. Define a function to be executed when the button is clicked
def on_button_clicked(b):
    with output_area:
        output_area.clear_output(wait=True) # Clear previous output before updating
        update_simulation() # Call the function to update simulation

# 5. Link the button's on_click event
weigh_button.on_click(on_button_clicked)

# 6. Display the button and the output widget
display(HBox([weigh_button, output_area2, output_area]))
```
