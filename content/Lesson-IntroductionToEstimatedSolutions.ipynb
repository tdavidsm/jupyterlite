# Weighing Out an Amount on a Digital Balance

We often have to weigh out a certain amount of some chemical or substance for a formula or recipe that we're trying to make. Let's take a look at an example of doing that. **Press play** on the code box below to run through the scenario.

```python
# @title
# prompt: Output text instructions that say, "Looking at your lab instructions, you see that you get to choose how much of chemical A to weigh out and you pick 10 grams. Then you see that you've got to get 1/3rd as much of chemical B. Use the button below to add more of chemical B to the scale and see what your sample weighs." Below this text should be a label titled, "Reading on Scale: " and then beside it is the outputted value of a variable titled currentMassOfChemicalB that starts at 0.00. Below this output is a button that says, "Add more of Chemical B" When this button is pressed, a random amount between 0.01 and 0.20 is added to the variable currentMassOfChemicalB. There is a second button that is labelled, "Ask the Teacher to See if You Got the Right Amount" When this button is clicked, the value of currentMassOfChemicalB is checked versus 3.33. If the variable value is within 0.2 of 3.33 then a string is output saying, "The teacher tells you that you did a good job and are close enough for this lab." If the variable value is outside of 0.2 away from 3.33, then the output string should read, "Nope! You're not close enough. Try again!" The value of currentMassOfChemicalB is then reset to 0.00 after a two second pause.

import time
import random
import ipywidgets as widgets
import IPython.display # Import IPython.display to explicitly call display

print("Looking at your lab instructions, you see that you get to choose how much of chemical A to weigh out. \n" +
     "You pick 10 grams.\n"+
     "Then you see that you've got to get 1/3rd as much of chemical B.\n"+
     "Use the button below to add more of chemical B to the scale and see what your sample weighs.\n"+
     "When you get to the right amount check with the teacher by pressing the button.\n\n")

currentMassOfChemicalB = 0.00

mass_output = widgets.Output()

with mass_output:
  print(f"Reading on Scale: {currentMassOfChemicalB:.2f} g")

IPython.display.display(mass_output) # Explicitly call display


def add_chemical_b(_):
  global currentMassOfChemicalB
  amount_to_add = random.uniform(0.01, 0.20)
  currentMassOfChemicalB += amount_to_add
  with mass_output:
    IPython.display.clear_output() # Explicitly call clear_output
    print(f"Reading on Scale: {currentMassOfChemicalB:.2f} g")


def check_amount(_):
  global currentMassOfChemicalB
  with mass_output:
    IPython.display.clear_output() # Explicitly call clear_output
    if abs(currentMassOfChemicalB - 3.33) <= 0.2:
      print("The teacher tells you that you did a good job and are close enough for this lab.")
    else:
      print("Nope! You're not close enough. Try again!")
      time.sleep(2)
      currentMassOfChemicalB = 0.00
      IPython.display.clear_output() # Explicitly call clear_output
      print(f"Reading on Scale: {currentMassOfChemicalB:.2f} g")

add_button = widgets.Button(description="Add more of Chemical B", layout=widgets.Layout(width='Auto', height='Auto'))
add_button.on_click(add_chemical_b)

check_button = widgets.Button(description="Ask the Teacher to See if You Got the Right Amount", layout=widgets.Layout(width='Auto', height='Auto'))
check_button.on_click(check_amount)

IPython.display.display(add_button) # Explicitly call display
IPython.display.display(check_button) # Explicitly call display
```

# Another Problem Closer to Chemistry

Imagine you've got to count a whole bunch of small things. This can be a real problem and can take a very long time. There are lots of ways around this, and often it involves estimation. For example, hit play on the code below and go through the process of figuring out how many points are on each graph. The first one you can easily count.

# Game 1

```python
# @title
# prompt: This code will generate three different graphs with three different amounts of points on them.
# One will have only a few and require exactly right answers, then the other two will have more and
# more points and require less accurate estimates, the last one will have a grid and suggest using
# sampling to guess the right number of points.

'''I am still working through getting this code to actually work right. I just tried a very vague prompt
with Gemini and got a lot of code that looks reasonable. I've beaten it into pretty good shape but have to
fix some formatting with where responses show up and it is tempting to divide the three "games" into
three different code blocks with some instructions between each.'''

# Import necessary libraries
import matplotlib.pyplot as plt
import numpy as np
from ipywidgets import IntText, Button, VBox
from IPython.display import display, clear_output

# Function to generate a scatter plot with a given number of points
def generate_scatter_plot(num_points, add_grid=False):
    plt.figure(figsize=(4, 4))
    x = np.random.rand(num_points) * 10
    y = np.random.rand(num_points) * 10
    plt.scatter(x, y, s=10)
    plt.xlim(0, 10)
    plt.ylim(0, 10)
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.title(f"Estimate the number of points!")

    if add_grid:
        plt.grid(True, linestyle='--', alpha=0.6)
        plt.xticks(np.arange(0, 11, 1))
        plt.yticks(np.arange(0, 11, 1))
        print("Hint: Try dividing the graph into sections and counting the points in a few sections to estimate.")

    plt.show()

# Function to handle the estimation game
def play_estimation_game(num_points, tolerance):
    generate_scatter_plot(num_points, add_grid=(num_points > 500))

    guess_input = IntText(description="Your guess:")
    check_button = Button(description="Check Guess")
    play_again_button = Button(description="Play Again", disabled=True)

    #display(VBox([guess_input, check_button])
    #clear_output(wait=True)

    display(VBox([guess_input, check_button]))

    def check_guess(b):
        #clear_output(wait=True)
        try:
            guess = guess_input.value
            if abs(guess - num_points) <= tolerance:
                print(f"Correct! The number of points was {num_points}.")
            else:
                print(f"Incorrect. The number of points was {num_points}. Your guess was off by {abs(guess - num_points)}.")

        except ValueError:
            print("Please enter a valid integer.")
        play_again_button.disabled = False
        display(play_again_button)

    check_button.on_click(check_guess)
    play_again_button.on_click(lambda b: play_estimation_game(num_points, tolerance))


# --- Game 1: Few points, exact answer needed ---
print("\n--- Game 1: Estimate the number of points ---")
print("For this graph, you'll need to count precisely.")
play_estimation_game(num_points=(3+np.random.randint(1,9)), tolerance=0)
```

# Game 2

For this next game, you might estimate or count. You can be off by up to 5 and your answer will be counted as correct.

```python
# @title
# --- Game 2: More points, small tolerance ---
print("\n--- Game 2: Estimate the number of points ---")
print("There are more points now. A small error is acceptable.")
play_estimation_game(num_points=(20+np.random.randint(1,30)), tolerance=5)
```

# Game 3

As you'll see when you hit play, this one has a lot more points! Counting just isn't very practical, and you're given even more leeway in putting in the answer. In fact, you just have to be within 100 of the right answer to get this counted right.

```python
# @title
# --- Game 3: Many points, larger tolerance, with grid ---
print("\n--- Game 3: Estimate the number of points ---")
print("There are many points on this graph. Estimation with sampling might be helpful.")
play_estimation_game(num_points=(300+np.random.randint(700)), tolerance=100)
```

Estimation works in the real world. We just have to understand how close we need to get to the "right" answer for our estimation to be "good enough."
